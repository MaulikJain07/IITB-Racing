```python
# libraries
import numpy as np
import scipy as sp
import scipy.special as spsp  #n choose k 
import matplotlib.pyplot as plt
```

```python
n = 40
p = 0.2
q = 1-p

# BINOMIAL DISTRIBUTION (EXACT)
PXbionomial = []
for k in range(n+1):
    probX = spsp.comb(n,k)*(p**k)*(q**(n-k))
    PXbionomial.append(probX)

# NORMAL DISTRIBUTION (APPROXIMATION)
mu = n*p
var = n*p*q
X = np.arange(-5,n,0.01) #creates an array from 0 to n with an increment of 0.01 between two consequtive elements.
PXnormal = 1/(np.sqrt(2*np.pi*var))*np.exp(-((X-mu)**2/(2*var)))

#plot
plt.figure(figsize=(12,7))
plt.plot(PXbionomial,label="Binomial")
plt.plot(X,PXnormal,label="Normal")
plt.legend()
plt.xlim(0,n)
plt.show()
```
